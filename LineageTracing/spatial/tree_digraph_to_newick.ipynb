```python
import networkx as nx

def digraph_to_newick(G, root=None):
    if root is None:
        roots = [n for n, d in G.in_degree() if d == 0]
        if len(roots) != 1:
            raise ValueError(f"no root: {roots}")
        root = roots[0]

    def build_newick(node):
        children = list(G.successors(node))
        if not children:  
            return str(node)
        else:  
            return "(" + ",".join(build_newick(c) for c in children) + ")" + str(node)

    return build_newick(root) + ";"
```

```python
G = TreeData_M1.obst['1_collapsed']
newick_str = digraph_to_newick(G)
with open("/syn1/liangzhen/jinhua_jilab_project/data/PEtracer/newick_trees/M1.1_collapsed.newick", "w") as f:
    f.write(newick_str)
```
